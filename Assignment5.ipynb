```python
#WordNet to identify semantic relationships such as synonymy, antonymy, and hypernymy in text data.
```

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.corpus import wordnet as wn
```

```python
# WordNet groups meanings into synsets (synonym sets).

wn.synsets("bank")
```

```python
# Find Synonyms
word = "happy"
synsets = wn.synsets(word)
synonyms = set()

for syn in synsets:
  for lemma in syn.lemmas():
    synonyms.add(lemma.name())
print("Synonyms of",word,":",synonyms)
```

```python
# find antonyms
word = "happy"

antonyms = set()

for syn in wn.synsets(word):
    for lemma in syn.lemmas():
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())

print("Antonyms of", word, ":", antonyms)
```

```python
# Find hypernyms (broader concepts)
# Hypernym = "is-a" relationship -> dog is a animal

word = "dog"

syn = wn.synsets(word)[0]   # pick first meaning

hypernyms = syn.hypernyms()

print("Hypernyms of", word)
for h in hypernyms:
    print("-", h.name(), ":", h.definition())
```

```python
# Find hyponyms(more specific concepts)
hyponyms = syn.hyponyms()

print("Hyponyms of", word)
for h in hyponyms[:5]:
    print("-", h.name())
```

```python
# Applying wordnet to real text
text = "Apple announced a major investment in artificial intelligence research."

words = text.lower().split()

for word in words:
    synsets = wn.synsets(word)
    if synsets:
        print(f"\nWord: {word}")
        print("Definition:", synsets[0].definition())
```
