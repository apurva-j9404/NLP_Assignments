```python
import nltk
from nltk.corpus import wordnet
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

# download resources
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
sentence = "I went to the bank to deposit money."
```

```python
#Tokenize Sentence
words = word_tokenize(sentence)
print(words)
```

```python
#Apply Word Sense Disambiguation
sense = lesk(words, 'bank')

print("Word:", 'bank')
print("Best Sense:", sense)
print("Definition:", sense.definition())
print("Example:", sense.examples())
```

```python
sentence2 = "The fisherman sat on the bank of the river."
words2 = word_tokenize(sentence2)

sense2 = lesk(words2, 'bank')

print("Word:", 'bank')
print("Best Sense:", sense2)
print("Definition:", sense2.definition())
```

```python
#WSD Function
def word_sense_disambiguation(sentence, target_word):

    words = word_tokenize(sentence)
    sense = lesk(words, target_word)

    if sense:
        print("Sentence:", sentence)
        print("Target Word:", target_word)
        print("Best Sense:", sense.name())
        print("Definition:", sense.definition())
        print("Examples:", sense.examples())
    else:
        print("No sense found")

# Test
word_sense_disambiguation(
    "The fisherman sat on the bank of the river",
    "bank"
)
```
