```python
import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = "NLP (Natural Language Processing) is an AI field enabling computers to understand, interpret, and generate human language (text and speech) by combining linguistics with computer science and machine learning, powering applications like voice assistants, translation, chatbots, and spam filters, to analyze data and automate tasks."
```

1. **Whitespace tokenization**

splits text by white spaces only.

```python
tokens_whitespace = text.split()
print(tokens_whitespace)
```

```python
#Punctuation-based Tokenization
from nltk.tokenize import wordpunct_tokenize
tokens_punct = wordpunct_tokenize(text)
print(tokens_punct)
```

```python
#Treebank Tokenizer
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokens_treebank = tokenizer.tokenize(text)
print(tokens_treebank)
```

```python
# Tweet tokenizer
from nltk.tokenize import TweetTokenizer
tokenizer = TweetTokenizer()
tokens_tweet = tokenizer.tokenize(text)
print(tokens_tweet)
```

```python
# MWE (Multi-Word Expression) Tokenizer
from nltk.tokenize import MWETokenizer
mwe = MWETokenizer([('machine', 'learning'), ('natural', 'language', 'processing')])
text2 = "I love machine learning and natural language processing"
tokens_mwe = mwe.tokenize(text2.split())
print(tokens_mwe)
```

```python
# STEMMING
```

```python
#Porter stemmer
from nltk.stem import PorterStemmer

ps = PorterStemmer()
words = ["running", "runs", "runner", "easily", "fairness"]

for word in words:
    print(word, "→", ps.stem(word))
```

```python
# Snowball stemmer
from nltk.stem import SnowballStemmer

ss = SnowballStemmer("english")

for word in words:
    print(word, "→", ss.stem(word))
```

```python
# LEMMATIZATION
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars", "studies"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

```python
lemmatizer.lemmatize("running", pos='v')  # verb
```
