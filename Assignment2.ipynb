```python
import nltk
nltk.download('punkt')
```

```python
documents = [
    "I love natural language processing",
    "Natural language processing is fun",
    "I love machine learning",
    "Machine learning is powerful"
]
```

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

tokenized_docs = [word_tokenize(doc.lower()) for doc in documents]
print(tokenized_docs)
```

```python
# # Bag of Words – Count Occurrence
# Bag of Words represents text as a vector of word frequencies, ignoring grammar and word order.
```

```python
from collections import Counter

bow_counts = [Counter(doc) for doc in tokenized_docs]
bow_counts
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X_counts = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out())
print(X_counts.toarray())
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(documents)

print("Features:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(X_tfidf.toarray())
```

```python
# Word2Vec Embeddings
from gensim.models import Word2Vec
```

```python
word2vec_model = Word2Vec(
    sentences=tokenized_docs,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)
```

```python

# get vector for word
vector = word2vec_model.wv["language"]
print("Vector length:", len(vector))
print(vector)
```

```python
# Find Similar Words
word2vec_model.wv.most_similar("learning")
```

```python
word2vec_model.wv.similarity("language", "processing")
```
