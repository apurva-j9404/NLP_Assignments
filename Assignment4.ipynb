```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Apple CEO Tim Cook visited India in 2024."
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```

```python
true_entities = [
    ("Apple", "ORG"),
    ("Tim Cook", "PERSON"),
    ("India", "GPE"),
    ("2024", "DATE")
]
```

```python
pred_entities = [(ent.text, ent.label_) for ent in doc.ents]

print("Predicted:", pred_entities)
print("Actual   :", true_entities)
```

```python
from sklearn.metrics import precision_recall_fscore_support, accuracy_score

y_true = [label for _, label in true_entities]
y_pred = [label for _, label in pred_entities]

precision, recall, f1, _ = precision_recall_fscore_support(
    y_true, y_pred, average="weighted"
)

accuracy = accuracy_score(y_true, y_pred)

print(f"Accuracy  : {accuracy:.2f}")
print(f"Precision : {precision:.2f}")
print(f"Recall    : {recall:.2f}")
print(f"F1 Score  : {f1:.2f}")
```
